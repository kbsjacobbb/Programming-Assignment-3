# A. POSITIONAL AND LABEL-BASED SLICING

# B. MODEL LOOKUP

```python
import pandas as pd
```

```python
cars = pd.read_csv('cars.csv')
```

```python
toyota_corolla = cars.iloc[(cars['Model'] == 'Toyota Corolla').values]
print("Toyota Corolla:")
print(toyota_corolla)
```

```python
pontiac_firebird = cars.iloc[(cars['Model'] == 'Pontiac Firebird').values][['Model', 'mpg', 'hp', 'wt']]
print("Pontiac Firebird:")
print(pontiac_firebird)
```

# C. MULTI-MODEL SUBSETTING

```python
import pandas as pd
```

```python
cars = pd.read_csv('cars.csv')
```

```python
models = ['Datsun 710', 'Lotus Europa', 'Ferrari Dino']
```

```python
selected_cars = cars.loc[cars['Model'].isin(models), ['Model', 'mpg', 'cyl', 'hp', 'gear']]
print(selected_cars)
print("Shape:", selected_cars.shape)
```
